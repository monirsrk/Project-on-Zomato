```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
url='/content/Zomato data .csv'
ds=pd.read_csv(url)
```

```python
ds.head(1)
```

```python
ds.describe()
```

```python
ds.info()
```

```python
ds['rate'] = ds['rate'].apply(lambda x: str(x).split('/')[0])
```

```python
ds.head()
```

```python
ds.isnull().sum()
```

```python
plt.figure(figsize=(12,5))
sns.countplot(x=ds['listed_in(type)'])
plt.show()
```

```python
groupby=ds.groupby('listed_in(type)')['votes'].sum()
result=pd.DataFrame(groupby)
plt.plot(result,c="red",marker="o")
```
